## 3 · Configuration

Larger price encoder: **d_model=768, n_layers=8, n_heads=12** (head_dim=64 unchanged).
Training extended to **10 000 steps** with val checkpoints every 500 steps.

**`RESUME=False`** — fresh training from random init (full LRs, Phase 1 warmup active).  
**`RESUME=True`** — warm-start from `nb03_best.pt`; LRs lowered 5×, Phase 1 skipped.

```python
import dataclasses as dc
import torch
from jepa_quant import JEPAConfig
from jepa_quant.config import (
    PriceEncoderConfig, PredictorConfig, DataConfig, TrainConfig
)

REGULARIZER = 'vicreg'  # 'vicreg' or 'codebook'
CONTEXT_LENGTH = 64
HORIZON        = 16
LATENT_DIM     = 256

# ── Resume flag ────────────────────────────────────────────────────────────
# False → train from scratch at full LRs with Phase 1 warmup.
# True  → warm-start from nb03_best.pt at 1/5× LRs, Phase 1 skipped.
RESUME = False

# LRs and warmup scale with RESUME so the same cell works for both modes.
_lr_proj    = 2e-5  if RESUME else 1e-4
_lr_adapter = 1e-5  if RESUME else 5e-5
_warmup     = 0     if RESUME else 500   # skip Phase 1 when resuming

cfg = JEPAConfig(
    # Larger price encoder: d_model 512→768, n_layers 6→8, n_heads 8→12
    price_encoder=PriceEncoderConfig(
        d_model=768,
        n_layers=8,
        n_heads=12,
        context_length=CONTEXT_LENGTH,
        latent_dim=LATENT_DIM,
    ),
    predictor=PredictorConfig(
        backend='lm' if USE_FOUNDATION_MODELS else 'transformer',
        latent_dim=LATENT_DIM,
    ),
    data=DataConfig(
        data_dir=DATA_DIR,
        context_length=CONTEXT_LENGTH,
        horizon=HORIZON,
    ),
    train=TrainConfig(
        max_steps=10000,
        warmup_proj_steps=_warmup,
        val_every=500,
        es_patience=6,
        es_min_delta=0.002,
        batch_size=256,
        lr_proj=_lr_proj,
        lr_adapter=_lr_adapter,
    ),
    regularizer=REGULARIZER,
)

print(cfg)
print(f'\nRESUME={RESUME}  lr_proj={_lr_proj:.0e}  lr_adapter={_lr_adapter:.0e}')
print(f'device: {torch.cuda.get_device_name(0) if torch.cuda.is_available() else "cpu"}')
```

## 4 · Data

```python
from jepa_quant.data.price_dataset import build_dataloaders

train_loader, val_loader = build_dataloaders(cfg)
print(f'Train batches: {len(train_loader)}  Val batches: {len(val_loader)}')
sample = next(iter(train_loader))
print(f'context: {sample["context"].shape}  target: {sample["target"].shape}')
```

## 5 · Build components & train

- Best `val_jepa` checkpoint is saved to Drive automatically after every improvement.
- Early stopping fires after 6 consecutive non-improving val checks (Δ < 0.002).
- If `RESUME=True`, weights are loaded from `nb03_best.pt` before training starts and
  the target encoder is re-synced to the loaded price encoder.

```python
import torch
from jepa_quant.training.trainer import build_components, JEPATrainer

CKPT_PATH = os.path.join(CKPT_DIR, 'nb03_best.pt')

# ── Load checkpoint if resuming ────────────────────────────────────────────
components = build_components(cfg)

if RESUME:
    if os.path.exists(CKPT_PATH):
        ckpt = torch.load(CKPT_PATH, map_location='cpu')
        components.price_encoder.load_state_dict(ckpt['price_encoder'])
        components.predictor.load_state_dict(ckpt['predictor'])
        # Keep target encoder in sync with the loaded price encoder weights.
        # TargetEncoder stores its copy in ._enc; update it directly so the
        # EMA starts from the right point rather than random init.
        components.target_encoder._enc.load_state_dict(ckpt['price_encoder'])
        print(f"Resumed from step {ckpt['step']}  val_jepa={ckpt['val_jepa']:.4f}")
    else:
        print(f"RESUME=True but {CKPT_PATH} not found — training from scratch")

# ── Checkpoint callback ────────────────────────────────────────────────────
def save_best(ckpt: dict) -> None:
    torch.save(ckpt, CKPT_PATH)
    print(f'  [ckpt] New best val_jepa={ckpt["val_jepa"]:.4f} at step {ckpt["step"]} → {CKPT_PATH}')

# ── Train ──────────────────────────────────────────────────────────────────
trainer = JEPATrainer(cfg, components, train_loader, val_loader, on_new_best=save_best)
history = trainer.train()
```

## 6 · Diagnostics

```python
import matplotlib.pyplot as plt

steps        = [r['step'] for r in history]
jepa_loss    = [r['jepa'] for r in history]
total_loss   = [r['loss'] for r in history]
val_steps    = [r['step'] for r in history if 'val_jepa' in r]
val_jepa     = [r['val_jepa'] for r in history if 'val_jepa' in r]

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

axes[0].plot(steps, total_loss, label='total loss', alpha=0.7)
axes[0].plot(steps, jepa_loss,  label='jepa loss',  alpha=0.7)
axes[0].set_title('Training loss')
axes[0].legend()
axes[0].set_xlabel('step')

if val_steps:
    axes[1].plot(val_steps, val_jepa, marker='o', label='val_jepa')
    axes[1].set_title('Validation JEPA loss')
    axes[1].set_xlabel('step')
    axes[1].legend()

# Regularization metrics (VICReg or codebook)
reg_keys = [k for k in history[0] if k.startswith('reg/')]
if reg_keys:
    for k in reg_keys:
        axes[2].plot(steps, [r.get(k, float('nan')) for r in history], label=k.split('/')[-1], alpha=0.8)
    axes[2].set_title('Regularization')
    axes[2].legend()
    axes[2].set_xlabel('step')
else:
    axes[2].set_visible(False)

plt.tight_layout()
plt.show()

print(f'Final step: {steps[-1]}  best val_jepa: {trainer._best_val_jepa:.4f}')
print(f'Checkpoint path: {CKPT_PATH}')
```
